```python
import polars as pl
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from typing import Dict, List
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pl.scan_parquet('../data_cleaning/cleaned_data/*.parquet')
```

```python
df.collect()
```

```python
# 'LazyFrame' does not have an 'n_unique' method; use 'n_unique' as an aggregation and collect the result.
num_sequences = df.select(pl.col("sequence_id").n_unique()).collect().item()
print(f"Number of unique sequences in filtered_df: {num_sequences}")
```

```python
df.describe()
```

```python
# Prepare columns of interest (excluding mean_temp, to be computed)
raw_columns = [
    "sequence_id", "cast_length_m", "rec_casting_speed",
    "width_pv", "deltatloose", "ar_to_sen_flow", "ar_to_sen_pressure",
    "tundish_pos_rel", "mould_level", "T_0", "T_1", "T_2", "T_3", "casting_program"
]
# "mean_temp" will be computed as the mean of T_0, T_1, T_2, T_3

# Check which requested columns exist in the DataFrame
columns_exist = [col for col in raw_columns if col in df.columns]
if "mould_level" not in columns_exist:
    print("Warning: 'mould_level' column not in DataFrame. Skipping it.")

# Check which temperature columns are available
temp_cols = [col for col in ["T_0", "T_1", "T_2", "T_3"] if col in columns_exist]
if not temp_cols:
    raise ValueError("No temperature columns found to compute mean_temp.")

# Collect only needed columns, compute mean_temp if possible
df_sel = df.select([pl.col(col) for col in columns_exist]).collect().to_pandas()

# Calculate mean_temp as the row-wise mean of available temperature columns
df_sel["mean_temp"] = df_sel[temp_cols].mean(axis=1)

# Prepare plot columns, replacing temp columns with mean_temp
columns_to_plot = [
    "sequence_id", "mean_temp", "cast_length_m", "rec_casting_speed",
    "width_pv", "deltatloose", "ar_to_sen_flow", "ar_to_sen_pressure",
    "tundish_pos_rel", "mould_level"
]
# Only keep those that are available (except mean_temp, which is now always present)
columns_exist_plot = ["mean_temp"] + [col for col in columns_to_plot if col in df_sel.columns and col != "mean_temp"]

# Group by sequence_id
if "sequence_id" not in df_sel.columns:
    raise ValueError("Column 'sequence_id' not found in selected DataFrame.")
grouped = df_sel.groupby("sequence_id")

# For each sequence_id, plot all variables (except sequence_id) as time series/subplots
for seq_id, subdf in grouped:
    cols_to_plot = [col for col in columns_exist_plot if col != "sequence_id"]
    n_vars = len(cols_to_plot)
    fig, axes = plt.subplots(n_vars, 1, figsize=(10, 2.5*n_vars), sharex=True)
    # Get casting_program for the sequence (assume all same in group, take first)
    casting_program = subdf["casting_program"].iloc[0] if "casting_program" in subdf.columns else "unknown"
    fig.suptitle(f"sequence_id: {seq_id} | casting_program: {casting_program}")

    if n_vars == 1:
        axes = [axes]

    for idx, col in enumerate(cols_to_plot):
        ax = axes[idx]
        ax.plot(subdf[col].to_numpy(), label=col)
        ax.set_ylabel(col)
        ax.legend(loc="upper right")

    plt.xlabel("Index")
    plt.tight_layout(rect=[0,0,1,0.97])
    plt.show()
```

```python
# List all sequence_ids in df, their max cast_length_m, their casting_program, and slabs_grade using Polars LazyFrame (df)
required_cols = ["sequence_id", "cast_length_m", "casting_program", "slabs_grade"]
for col in required_cols:
    if col not in df.columns:
        raise ValueError(f"Column '{col}' must be present in df to include it.")

if isinstance(df, pl.LazyFrame):
    out = (
        df
        .group_by("sequence_id")
        .agg([
            pl.col("cast_length_m").max().alias("max_cast_length"),
            pl.col("casting_program").first().alias("casting_program"),
            pl.col("slabs_grade").first().alias("slabs_grade")
        ])
        .sort("sequence_id")
        .collect()
    )
else:
    out = (
        df
        .groupby("sequence_id")
        .agg([
            pl.col("cast_length_m").max().alias("max_cast_length"),
            pl.col("casting_program").first().alias("casting_program"),
            pl.col("slabs_grade").first().alias("slabs_grade")
        ])
        .sort("sequence_id")
    )

```

```python
# Show rows from 'out' where sequence_id is 53912, 53916, 53940, or 54693
filtered_out = out.filter(pl.col("sequence_id").is_in([53912, 53916, 53940, 54693]))
print(filtered_out)
```
