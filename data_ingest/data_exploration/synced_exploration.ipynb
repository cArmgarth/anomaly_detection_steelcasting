```python
import polars as pl
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pl.scan_parquet('../data_cleaning/cleaned_data/*.parquet')
```

```python
df.collect()
```

```python
# Plot sequence_id over time
df_seq = df.select(['datetime', 'sequence_id']).collect()

plt.figure(figsize=(14, 5))
plt.plot(
    df_seq['datetime'].to_pandas(),
    df_seq['sequence_id'].to_pandas(),
    linestyle='none',
    marker='.', 
    alpha=0.5
)
plt.xlabel("Time")
plt.ylabel("sequence_id")
plt.title("sequence_id Over Time")
plt.tight_layout()
plt.show()
```

```python
# For all sequence_ids, plot the mean temp of all T_ columns over time, and include slabs_grade in the legend

# Find all T_ columns
schema = df.schema
t_cols = [col for col in schema if col.startswith('T_')]

# Collect necessary columns: datetime, sequence_id, slabs_grade, and all T_ cols
cols_to_select = ['datetime', 'sequence_id', 'slabs_grade'] + t_cols
df_temp = df.select(cols_to_select).collect()

# Calculate the mean temperature per row (across all T_ columns)
df_temp = df_temp.with_columns(
    pl.concat_list(t_cols).list.eval(pl.element().cast(pl.Float64)).list.mean().alias('mean_temp')
)

# For each sequence_id, get its slabs_grade (take first non-null if ambiguous)
slabs_grade_per_seq = (
    df_temp
    .group_by("sequence_id")
    .agg(pl.col("slabs_grade").filter(pl.col("slabs_grade").is_not_null()).first().alias("slabs_grade"))
    .to_pandas()
    .set_index("sequence_id")["slabs_grade"]
)

# Group by datetime and sequence_id and compute mean_temp
df_plot = (
    df_temp
    .group_by(['datetime', 'sequence_id'])
    .agg(pl.col('mean_temp').mean())
    .sort('datetime')
)

# Convert to pandas for plotting
df_plot_pd = df_plot.to_pandas()

plt.figure(figsize=(16, 6))
for seq_id, grp in df_plot_pd.groupby('sequence_id'):
    # Get slabs_grade for this sequence_id, handle missing gracefully
    slabs_grade = slabs_grade_per_seq.get(seq_id, "Unknown")
    label = f"sequence_id {int(seq_id)} (slabs_grade {int(slabs_grade) if pd.notnull(slabs_grade) else 'Unknown'})"
    plt.plot(grp['datetime'], grp['mean_temp'], label=label, alpha=0.7)

plt.xlabel("Time")
plt.ylabel("Mean temp (across all T_ cols)")
plt.title("Mean Temperature per sequence_id Over Time")
# Set legend as vertical single column list
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', fontsize='small', ncol=1, borderaxespad=0.)
plt.tight_layout()
plt.show()
```

```python
# Plot cast_length_m over time
# The correct column name (according to your schema) is 'cast_length_m', not 'cast_length'

# Collect necessary columns: datetime, sequence_id, cast_length_m
cols_cast_length = ['datetime', 'sequence_id', 'cast_length_m']
df_cast_length = df.select(cols_cast_length).collect()

# Group by datetime and sequence_id, and aggregate (mean in case of duplicates)
df_cast_plot = (
    df_cast_length
    .group_by(['datetime', 'sequence_id'])
    .agg(pl.col('cast_length_m').mean().alias('cast_length_m'))
    .sort('datetime')
)

# Convert to pandas for plotting
df_cast_plot_pd = df_cast_plot.to_pandas()

plt.figure(figsize=(16, 6))
for seq_id, grp in df_cast_plot_pd.groupby('sequence_id'):
    plt.plot(grp['datetime'], grp['cast_length_m'], label=f"sequence_id {int(seq_id)}", alpha=0.7)

plt.xlabel("Time")
plt.ylabel("Cast Length (m)")
plt.title("Cast Length per sequence_id Over Time")
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', fontsize='small', ncol=1, borderaxespad=0.)
plt.tight_layout()
plt.show()
```

```python
df_pd = df.collect().to_pandas()
```

```python
# For each sequence, plot mean temp and cast length over time as subplots

import matplotlib.pyplot as plt

# Identify the relevant T_ columns for temperature
temp_cols = [col for col in df_pd.columns if col.startswith("T_")]

# Get the unique sequence ids
sequence_ids = df_pd["sequence_id"].unique()

# Generate one subplot set per sequence
n_seq = len(sequence_ids)
fig, axes = plt.subplots(n_seq, 2, figsize=(16, 5 * n_seq), sharex=False)

# If only one sequence, axes may not be a 2D array
if n_seq == 1:
    axes = axes.reshape(1, -1)

for idx, seq_id in enumerate(sequence_ids):
    df_seq = df_pd[df_pd["sequence_id"] == seq_id].copy()
    # Sort by datetime
    df_seq = df_seq.sort_values("datetime")

    # Compute mean temp row-wise (across all T_ cols)
    df_seq["mean_temp"] = df_seq[temp_cols].mean(axis=1)
    # First subplot: mean temp
    ax1 = axes[idx, 0]
    ax1.plot(df_seq["datetime"], df_seq["mean_temp"], color='tab:red')
    ax1.set_title(f"Sequence {seq_id}: Mean Temperature Over Time")
    ax1.set_ylabel("Mean Temperature")
    ax1.set_xlabel("Time")

    # Second subplot: cast length
    ax2 = axes[idx, 1]
    ax2.plot(df_seq["datetime"], df_seq["cast_length_m"], color='tab:blue')
    ax2.set_title(f"Sequence {seq_id}: Cast Length Over Time")
    ax2.set_ylabel("Cast Length (m)")
    ax2.set_xlabel("Time")

plt.tight_layout()
plt.show()
```
