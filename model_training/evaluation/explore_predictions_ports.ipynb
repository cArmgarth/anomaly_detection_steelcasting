```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.graph_objs as go
```

```python
df = pd.read_csv("../models/test_set_anomalies_with_ports.csv")
```

```python
df
```

```python
# List unique sequence_ids in the dataset
unique_sequences = df['sequence_id'].unique()
print("Unique sequence_ids:", unique_sequences)
```

```python
# Create a dataframe for sequence 53912
df_53912 = df[df['sequence_id'] == 53912]
df_53916 = df[df['sequence_id'] == 53916]
df_53940 = df[df['sequence_id'] == 53940]
df_54693 = df[df['sequence_id'] == 54693]
```

```python
df_53912
```

```python
# Prepare data
cycle_times_53912 = pd.to_datetime(df_53912['cycle_datetime'])
anomaly_scores_53912 = df_53912['anomaly_score']

# Prepare hover text with top features, their deviations, and port_info
hover_text_53912 = (
    "Cycle: " + df_53912['cycle_id'].astype(str) + "<br>" +
    "Datetime: " + df_53912['cycle_datetime'].astype(str) + "<br>" +
    "Anomaly Score: " + df_53912['anomaly_score'].round(2).astype(str) + "<br>" +
    "Top Features:<br>" +
    df_53912[['top_feature_1', 'top_feature_2', 'top_feature_3', 'top_feature_4', 'top_feature_5', 
              'top_feature_1_deviation', 'top_feature_2_deviation', 'top_feature_3_deviation', 
              'top_feature_4_deviation', 'top_feature_5_deviation', 'port_info']]
    .apply(lambda row: 
        f"{row['top_feature_1']}: {row['top_feature_1_deviation']:.2f}<br>" +
        f"{row['top_feature_2']}: {row['top_feature_2_deviation']:.2f}<br>" +
        f"{row['top_feature_3']}: {row['top_feature_3_deviation']:.2f}<br>" +
        f"{row['top_feature_4']}: {row['top_feature_4_deviation']:.2f}<br>" +
        f"{row['top_feature_5']}: {row['top_feature_5_deviation']:.2f}<br>" +
        f"Ports: {row['port_info'] if pd.notnull(row['port_info']) else 'N/A'}"
    , axis=1)
)

# Create the Plotly figure
fig_53912 = go.Figure()

fig_53912.add_trace(go.Scatter(
    x=cycle_times_53912,
    y=anomaly_scores_53912,
    mode='lines+markers',
    marker=dict(size=7),
    text=hover_text_53912,
    hoverinfo='text'
))

fig_53912.update_layout(
    title='Anomaly Score for Sequence 53912',
    xaxis_title='Cycle Datetime',
    yaxis_title='Anomaly Score',
    hovermode='closest',
    width=1200,
    height=500
)

# Flip the anomaly score scale so higher values indicate higher anomaly likelihood
fig_53912.update_yaxes(autorange='reversed')

fig_53912.show()
```
